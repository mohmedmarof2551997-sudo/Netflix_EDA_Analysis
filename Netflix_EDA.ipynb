```python
# Check for missing values
print('❌ Missing Values in Each Column:')
null_counts = df.isnull().sum()
print(null_counts)
```

## Convert to datetime 

```python
df['date_added']  =  pd.to_datetime(df['date_added'].str.strip(), format='%B %d, %Y', errors='coerce')
df['Year']         = df['date_added'].dt.year
df['Month']        = df['date_added'].dt.month
df['MonthName']    = df['date_added'].dt.strftime('%b')
df['Quarter']       = df['date_added'].dt.quarter.map({1:'Q1',2:'Q2',3:'Q3',4:'Q4'})
df['YearMonth']    = df['date_added'].dt.to_period('M')
df['DayOfWeek']    = df['date_added'].dt.day_name()
```

## FILL MISSING VALUE 

```python
for col in ['director','cast','country','rating']:
    df[col] = df[col].fillna('Unknown')
```

## Duration split

```python
df[['num','unit']] = df['duration'].str.extract(r'(\d+)\s*([A-Za-z]+)')
df['num'] = pd.to_numeric(df['num'], errors='coerce')
df['unit'] = (
    df['unit'].str.lower().str.strip().replace({'season':'seasons'})) 
df['movie_minutes'] = df['num'].where(df['unit']=='min')
df['tv_seasons']    = df['num'].where(df['unit']=='seasons')
```

## COLUMN CREATED 

```python
df['content_age']            = 2021 - df['release_year']
df['content_age_when_added'] = df['Year'] - df['release_year']
df['genre_count']            = df['listed_in'].str.count(',') + 1
df['is_multi_genre']         = (df['genre_count'] > 1).astype(int)
df['is_mature']              = df['rating'].isin(['TV-MA','R','NC-17']).astype(int)
df['is_us_content']          = df['country'].str.contains('United States', na=False).astype(int)
df['is_international']       = (~df['country'].str.contains('United States', na=False) & (df['country'] != 'Unknown')).astype(int)
df['cast_count'] = (df['cast'].fillna('Unknown').apply(lambda x: len(x.split(',')) if x != 'Unknown' else 0))
df['description_length']     = df['description'].fillna('').str.len()
```

## CUT THE YEAR INTO Era

```python
bins = [0, 1999, 2009, 2015, 2018, 2100]
labels = ['Classic','2000s','2010-15','2016-18','2019+']

df['content_era'] = pd.cut(df['release_year'], bins=bins, labels=labels)
```

## MASTER KPI DASHBOARD — 19 KPIs

```python
movies = df[df['type']=='Movie']
tv= df[df['type']=='TV Show']
total= len(df)
n_movies= len(movies)
n_tv= len(tv)
n_genres= len(set(df['listed_in'].dropna().str.split(', ').sum()))
n_countries= len(set(country.strip()
for sublist in df['country'].dropna().str.split(',')
for country in sublist
if country.strip() != '' and country.strip() != 'Unknown' ))
n_directors = len(set(director.strip()
for sublist in df[df['director'] != 'Unknown']['director'].dropna().str.split(',')
for director in sublist
if director.strip() != ''))
pct_mature = df['is_mature'].mean()*100
pct_intl = df['is_international'].mean()*100
med_movie_dur = movies['movie_minutes'].median()
med_tv_seasons = tv['tv_seasons'].median()
peak_year  = df.groupby('Year').size().idxmax()
single_season = (tv['tv_seasons']==1).mean()*100
countries = pd.Series([country.strip() 
for countries in df['country'].dropna() 
for country in countries.split(',')
if country.strip() and country.strip() != 'Unknown'])
top_country = countries.value_counts().index[0]
genres = pd.Series([genre.strip().title()
for genres in df['listed_in'].dropna()
for genre in genres.split(',')
if genre.strip() and genre.strip().lower() != 'unknown'])
top_genre =genres.value_counts().index[0]
top_rating = df[df['rating']!='Unknown']['rating'].value_counts().index[0]
med_cast = df[df['cast_count']>0]['cast_count'].median()
pct_us = df['is_us_content'].mean()*100
med_age_added = df['content_age_when_added'].median()
year_counts = df.groupby('Year').size().sort_index()
yoy_growth = year_counts.pct_change() * 100
```

```python
def print_kpi(title, value, suffix=""):
    print(f"{title:<15}: {value} {suffix}")

print("\n" + "="*60)
print("🎬 NETFLIX ANALYTICS KPI")
print("="*60)

print_kpi("Total Content", total)
print_kpi("Movies", n_movies)
print_kpi("TV Shows", n_tv)
print_kpi("Genres", n_genres)
print_kpi("Countries", n_countries)
print_kpi("Directors", n_directors)
print_kpi("peak year",int(peak_year))


print("-"*60)

print_kpi("Mature Content", f"{pct_mature:.2f}", "%")
print_kpi("International Content", f"{pct_intl:.2f}", "%")
print_kpi("US Content", f"{pct_us:.2f}", "%")
print_kpi("Peak YoY Growth %", f"{yoy_growth.loc[peak_year]}", "%")


print("-"*60)

print_kpi("median Movie Duration", f"{med_movie_dur:.2f}", "min")
print_kpi("median TV Seasons", f"{med_tv_seasons:.2f}")
print_kpi("Single Season Shows", f"{single_season:.2f}", "%")
print_kpi("median cast size", f"{med_cast:.2f}", "cast")
print_kpi("median age added", f"{med_age_added:.2f}", "yr")

print("-"*60)

print_kpi("Top Country", top_country)
print_kpi("Top Genre", top_genre)
print_kpi("Top Rating", top_rating)

print("="*60)
```
